```python
import ast
import sys
from pathlib import Path

import pandas as pd

# Add project root to Python path
project_root = Path.cwd().parent  # Go up one level from current directory
sys.path.append(str(project_root))

# from utils.task_preps import load_reviews_data
from llm_providers.openai_client.data_prep_for_llm import filter_reviews_data


# load_dotenv(dotenv_path=os.path.join(project_root, '.env'), override=True)
def load_reviews_data(file_path: str) -> pd.DataFrame:
    # Read the CSV file with semicolon separator
    df = pd.read_csv(file_path, sep=";")

    # Convert string representations of lists to actual lists
    df["ai_positive_topics"] = df["ai_positive_topics"].apply(ast.literal_eval)
    df["ai_negative_topics"] = df["ai_negative_topics"].apply(ast.literal_eval)

    return df


file_path = "./data/results/gpt-4.1-nano/reviews_2025_02_22_210638_com.gardrops_gpt-4.1-nano.csv"

# Load reviews data
data_reviews = load_reviews_data(file_path)

# filter reviews data
data_reviews, num_of_reviews = filter_reviews_data(data_reviews)
data_reviews
```

```python
# load_dotenv(dotenv_path=os.path.join(project_root, '.env'), override=True)
def load_reviews_data_gemini(file_path: str) -> pd.DataFrame:
    # Read the CSV file with semicolon separator
    df = pd.read_csv(file_path, sep="¤", engine="python")

    # Convert string representations of lists to actual lists
    df["ai_positive_topics"] = df["ai_positive_topics"].apply(ast.literal_eval)
    df["ai_negative_topics"] = df["ai_negative_topics"].apply(ast.literal_eval)

    return df
```

```python
# load google gemini outputs
file_path_gemini = "./data/results/gemini-2.0-flash/reviews_2025_02_22_210638_com.gardrops_gemini-2.0-flash.csv"
df_gemini = load_reviews_data_gemini(file_path_gemini)
df_gemini
```
